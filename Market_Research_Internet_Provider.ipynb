# Analisis Tren

```python
from google.colab import drive
import pandas as pd

for sheet_name in xls.sheet_names:
    df = pd.read_excel(xls, sheet_name=sheet_name)

    if 'datetime' in df.columns and 'sentiment' in df.columns:
        df['datetime'] = pd.to_datetime(df['datetime'])
        df['bulan'] = df['datetime'].dt.to_period('M')

        df = df[df['datetime'].dt.month.isin([7, 8, 9, 10, 11, 12])]

        percakapan_per_bulan = df.groupby('bulan').size().reset_index(name='jumlah_percakapan')
        print(f"Tabel Tren Percakapan per Bulan - {sheet_name}")
        display(percakapan_per_bulan)
        print("\n" + "-"*50 + "\n")

        sentimen_per_bulan = df.groupby(['bulan', 'sentiment']).size().unstack().fillna(0).reset_index()
        print(f"Tabel Sentimen per Bulan - {sheet_name}")
        display(sentimen_per_bulan)
        print("\n" + "="*50 + "\n")
    else:
        print(f"Sheet {sheet_name} tidak memiliki kolom 'datetime' atau 'sentiment'.")
```

# Distribusi Channel

```python
import pandas as pd

df = pd.read_excel(xls, sheet_name="All Data")
if df.shape[1] >= 5:
    df.columns = list(df.columns)
    df = df.iloc[:, [4, 0, 1] + list(range(5, df.shape[1]))]
    df.columns = ['date', 'brand', 'service'] + list(df.columns[3:])

    df['date'] = pd.to_datetime(df['date'])
    df['bulan'] = df['date'].dt.month
    df['tahun'] = df['date'].dt.year

    df = df[(df['tahun'] == 2024) & (df['bulan'].between(7, 12))]

    bulan_dict = {7: "Juli", 8: "Agustus", 9: "September", 10: "Oktober", 11: "November", 12: "Desember"}
    bulan_kategori = pd.Categorical(df['bulan'].map(bulan_dict),
                                    categories=["Juli", "Agustus", "September", "Oktober",
                                                "November", "Desember"],
                                    ordered=True)
    df['bulan_kategori'] = bulan_kategori

    distribusi_channel_all = df['service'].value_counts().reset_index()
    distribusi_channel_all.columns = ['channel', 'jumlah_percakapan']
    print(f"Distribusi Channel All Periode (Juli - Desember 2024) - All Data")
    display(distribusi_channel_all)
    print("\n" + "-"*50 + "\n")

    distribusi_channel_bulanan = df.groupby(['service', 'bulan_kategori'], observed=True).size().unstack().fillna(0).reset_index()
    distribusi_channel_bulanan = distribusi_channel_bulanan.sort_index(axis=1, key=lambda x: x.map(lambda y: bulan_kategori.categories.get_loc(y) if y in bulan_kategori.categories else float('inf')))

    print(f"Distribusi Channel Per Bulan (Juli - Desember 2024) - All Data")
    display(distribusi_channel_bulanan)
    print("\n" + "="*50 + "\n")

    distribusi_channel_brand = df.groupby(['brand', 'service'], observed=True).size().unstack().fillna(0).reset_index()
    print(f"Distribusi Channel Per Brand (Juli - Desember 2024) - All Data")
    display(distribusi_channel_brand)
    print("\n" + "="*50 + "\n")

else:
    print(f"Sheet 'All Data' memiliki kurang dari 5 kolom, tidak bisa mengambil date, brand, atau service.")
```

```python
import pandas as pd

xls = pd.ExcelFile(file_path)

keywords_gangguan = ["ganggu", "lemot", "loss", "ilang", "los", "bermasalah", "kendala", "mati", "ngelag", "jelek", "gangguan", "down", "trouble", "ngadat", "gabisa", "kenapa", "busuk", "susah", "lelet", "bapuk", "putus", "hilang", "error", "lag"]
keywords_harga_paket = ["kecewa", "mahal", "mbps", "pindah", "rugi", "naik", "ganti", "mbs", "biaya", "tagihan"]
keywords_layanan_pelanggan = ["respon", "repot", "keluhan", "dipersulit", "ribet", "komplain", "lama", "kendala", "solusi", "benerin", "pengaduan", "teknisi", "tolol", "complain", "lapor", "pelaporan", "tanggapan"]

sheet_name = xls.sheet_names[0]
df = pd.read_excel(xls, sheet_name=sheet_name)


if df.shape[1] >= 23:
    df['kolom_23'] = df.iloc[:, 22].astype(str)

    if 'sentiment' in df.columns:
        df = df[df['sentiment'].str.lower() == 'negative']

    if 'datetime' in df.columns:
        df['datetime'] = pd.to_datetime(df['datetime'], errors='coerce')
        df = df[df['datetime'].dt.month.isin([7, 8, 9, 10, 11, 12])]

    df['keluhan_gangguan'] = df['kolom_23'].str.contains('|'.join(keywords_gangguan), case=False, na=False)
    df['keluhan_harga_paket'] = df['kolom_23'].str.contains('|'.join(keywords_harga_paket), case=False, na=False)
    df['keluhan_layanan_pelanggan'] = df['kolom_23'].str.contains('|'.join(keywords_layanan_pelanggan), case=False, na=False)

    jumlah_keluhan_gangguan = df['keluhan_gangguan'].sum()
    jumlah_keluhan_harga_paket = df['keluhan_harga_paket'].sum()
    jumlah_keluhan_layanan_pelanggan = df['keluhan_layanan_pelanggan'].sum()


    results_df = pd.DataFrame({
        'Worksheet': [sheet_name],
        'Jumlah Keluhan Gangguan Jaringan': [jumlah_keluhan_gangguan],
        'Jumlah Keluhan Harga & Paket Internet': [jumlah_keluhan_harga_paket],
        'Jumlah Keluhan Layanan Pelanggan': [jumlah_keluhan_layanan_pelanggan]
    })

    display(results_df)
else:
    print(f"Sheet {sheet_name} tidak memiliki minimal 23 kolom.")
```

```python
import pandas as pd

xls = pd.ExcelFile(file_path)

keywords_gangguan = ["ganggu", "lemot", "loss", "ilang", "los", "bermasalah", "kendala", "mati", "ngelag", "jelek", "gangguan", "down", "trouble", "ngadat", "gabisa", "kenapa", "busuk", "susah", "lelet", "bapuk", "putus", "hilang", "error", "lag"]
keywords_harga_paket = ["kecewa", "mahal", "mbps", "pindah", "rugi", "naik", "ganti", "mbs", "biaya", "tagihan"]
keywords_layanan_pelanggan = ["respon", "repot", "keluhan", "dipersulit", "ribet", "komplain", "lama", "kendala", "solusi", "benerin", "pengaduan", "teknisi", "tolol", "complain", "lapor", "pelaporan", "tanggapan"]

sheet_name = xls.sheet_names[0]
df = pd.read_excel(xls, sheet_name=sheet_name)

if df.shape[1] >= 23:
    df['kolom_23'] = df.iloc[:, 22].astype(str)

    if 'sentiment' in df.columns:
        df = df[df['sentiment'].str.lower() == 'negative']

    if 'datetime' in df.columns:
        df['datetime'] = pd.to_datetime(df['datetime'], errors='coerce')
        df = df[df['datetime'].dt.month.isin([7, 8, 9, 10, 11, 12])]

    if df.shape[1] >= 2:
        df['brand'] = df.iloc[:, 0].astype(str)
        df['channel'] = df.iloc[:, 1].astype(str)

        results_channel = []
        for channel, group in df.groupby('channel'):
            keluhan_gangguan = group['kolom_23'].str.contains('|'.join(keywords_gangguan), case=False, na=False).sum()
            keluhan_harga_paket = group['kolom_23'].str.contains('|'.join(keywords_harga_paket), case=False, na=False).sum()
            keluhan_layanan_pelanggan = group['kolom_23'].str.contains('|'.join(keywords_layanan_pelanggan), case=False, na=False).sum()

            results_channel.append({
                'Channel': channel,
                'Jumlah Keluhan Gangguan Jaringan': keluhan_gangguan,
                'Jumlah Keluhan Harga & Paket Internet': keluhan_harga_paket,
                'Jumlah Keluhan Layanan Pelanggan': keluhan_layanan_pelanggan
            })

        results_channel_df = pd.DataFrame(results_channel)
        results_channel_df.loc['Total'] = results_channel_df.iloc[:, 1:].sum()
        results_channel_df.loc['Total', 'Channel'] = 'Total'

        results_brand = []
        for brand, group in df.groupby('brand'):
            keluhan_gangguan = group['kolom_23'].str.contains('|'.join(keywords_gangguan), case=False, na=False).sum()
            keluhan_harga_paket = group['kolom_23'].str.contains('|'.join(keywords_harga_paket), case=False, na=False).sum()
            keluhan_layanan_pelanggan = group['kolom_23'].str.contains('|'.join(keywords_layanan_pelanggan), case=False, na=False).sum()

            results_brand.append({
                'Brand': brand,
                'Jumlah Keluhan Gangguan Jaringan': keluhan_gangguan,
                'Jumlah Keluhan Harga & Paket Internet': keluhan_harga_paket,
                'Jumlah Keluhan Layanan Pelanggan': keluhan_layanan_pelanggan
            })

        results_brand_df = pd.DataFrame(results_brand)
        results_brand_df.loc['Total'] = results_brand_df.iloc[:, 1:].sum()
        results_brand_df.loc['Total', 'Brand'] = 'Total'

        print("Distribusi Jenis Keluhan per Channel:")
        display(results_channel_df)
        print("\nDistribusi Keluhan per Brand:")
        display(results_brand_df)
    else:
        print(f"Sheet {sheet_name} tidak memiliki minimal 2 kolom untuk brand dan channel.")
else:
    print(f"Sheet {sheet_name} tidak memiliki minimal 23 kolom.")
```

```python
import pandas as pd

# Membaca file Excel
xls = pd.ExcelFile(file_path)

# Kata kunci untuk kategori keluhan
keywords_edukasi = ["edukasi teknologi", "literasi digital", "tips teknologi", "panduan teknologi", "cara pakai aplikasi", "cara pakai fitur", "belajar teknologi", "belajar digital", "paham teknologi", "info teknologi", "info digital", "teknologi terkini", "update teknologi", "fitur teknologi", "kenali fitur", "manfaat teknologi", "cara kerja teknologi", "teknologi sehari-hari", "penggunaan teknologi", "gaya hidup digital", "internet sehat"]
keywords_produk = ["produk baru", "layanan internet", "fitur layanan", "paket internet", "pilihan paket", "upgrade layanan", "kecepatan internet", "teknologi terbaru", "fitur unggulan", "detail produk", "layanan tambahan", "info paket", "kuota internet", "modem wifi", "perangkat internet"]
keywords_support = ["layanan pelanggan", "customer service", "hubungi cs", "komplain layanan", "solusi cepat", "kendala jaringan", "respon lambat", "tips penggunaan", "cara pakai layanan", "panduan teknis", "cara setting", "bantuan teknis", "teknisi datang", "trouble internet", "solusi masalah", "respon", "tanggapi"]
keywords_promo = ["promo spesial", "diskon langganan", "cashback", "gratis kuota", "voucher internet", "penawaran menarik", "promo terbatas", "bonus kuota", "kode promo", "promo paket", "potongan harga", "event spesial", "bundling promo", "flash sale", "hadiah langsung", "promo", "diskon", "cashback"]
keywords_testimoni = ["testimoni", "pengalaman", "review", "cerita", "ulasan", "komentar positif", "feedback pelanggan", "kepuasan pengguna", "cerita nyata", "pengalaman buruk", "rekomendasi teman", "opini pengguna", "pengguna setia", "komentar jujur"]
keywords_lifestyle = ["gaya hidup", "streaming", "gaming", "hiburan", "nonton film", "work from home", "belajar online", "main game", "musik" , "tren digital" , "aktivitas online"]

# Membaca sheet pertama
sheet_name = xls.sheet_names[0]
df = pd.read_excel(xls, sheet_name=sheet_name)

# Filter hanya sentimen "positive" dan "neutral"
if 'sentimen' in df.columns:
    df = df[df['sentimen'].isin(['positive', 'neutral'])]

# Pastikan dataset memiliki kolom minimal 23
if df.shape[1] >= 23:
    df['kolom_23'] = df.iloc[:, 22].astype(str)

    # Filter data berdasarkan bulan (opsional)
    if 'datetime' in df.columns:
        df['datetime'] = pd.to_datetime(df['datetime'])
        df = df[df['datetime'].dt.month.isin([7, 8, 9, 10, 11, 12])]

    # Pastikan ada minimal dua kolom untuk brand dan channel
    if df.shape[1] >= 2:
        df['brand'] = df.iloc[:, 0].astype(str)
        df['channel'] = df.iloc[:, 1].astype(str)

        results_brand = []
        for brand, group in df.groupby('brand'):
            keluhan_edukasi = group['kolom_23'].apply(lambda x: any(word in x.lower() for word in keywords_edukasi)).sum()
            keluhan_produk = group['kolom_23'].apply(lambda x: any(word in x.lower() for word in keywords_produk)).sum()
            keluhan_support = group['kolom_23'].apply(lambda x: any(word in x.lower() for word in keywords_support)).sum()
            keluhan_promo = group['kolom_23'].apply(lambda x: any(word in x.lower() for word in keywords_promo)).sum()
            keluhan_testimoni = group['kolom_23'].apply(lambda x: any(word in x.lower() for word in keywords_testimoni)).sum()
            keluhan_lifestyle = group['kolom_23'].apply(lambda x: any(word in x.lower() for word in keywords_lifestyle)).sum()

            results_brand.append({
                'Brand': brand,
                'Edukasi Teknologi & Digital': keluhan_edukasi,
                'Produk & Layanan': keluhan_produk,
                'Customer Support & Tips Pemakaian': keluhan_support,
                'Promo & Penawaran Khusus': keluhan_promo,
                'Testimoni & Cerita Pelanggan': keluhan_testimoni,
                'Lifestyle & Entertainment': keluhan_lifestyle
            })

        # Buat DataFrame hasil
        results_brand_df = pd.DataFrame(results_brand)
        results_brand_df.loc['Total'] = results_brand_df.iloc[:, 1:].sum()
        results_brand_df.loc['Total', 'Brand'] = 'Total'

        # Tampilkan hasil
        print("Distribusi Jenis Keluhan per Brand (Sentimen Positive & Neutral):")
        display(results_brand_df)
    else:
        print(f"Sheet {sheet_name} tidak memiliki minimal 2 kolom untuk brand dan channel.")
else:
    print(f"Sheet {sheet_name} tidak memiliki minimal 23 kolom.")
```

⁠Rata-Rata engagement & Presentasenya per content pillar

```python
import pandas as pd
import numpy as np

df['engagement'] = pd.to_numeric(df.iloc[:, 6], errors='coerce')

engagement_values = []
total_keluhan_values = []

for brand in results_brand_df['Brand']:
    if brand == 'Total':
        continue

    group = df[df['brand'] == brand]

    def hitung_engagement(keywords):
        mask = group['kolom_23'].str.contains('|'.join(keywords), case=False, na=False)
        return group.loc[mask, 'engagement'].sum(), mask.sum()

    eng_edukasi, total_edukasi = hitung_engagement(keywords_edukasi)
    eng_produk, total_produk = hitung_engagement(keywords_produk)
    eng_support, total_support = hitung_engagement(keywords_support)
    eng_promo, total_promo = hitung_engagement(keywords_promo)
    eng_testimoni, total_testimoni = hitung_engagement(keywords_testimoni)
    eng_lifestyle, total_lifestyle = hitung_engagement(keywords_lifestyle)

    total_engagement = eng_edukasi + eng_produk + eng_support + eng_promo + eng_testimoni + eng_lifestyle
    total_keluhan = total_edukasi + total_produk + total_support + total_promo + total_testimoni + total_lifestyle

    engagement_values.append(total_engagement)
    total_keluhan_values.append(total_keluhan)

results_brand_df.loc[results_brand_df['Brand'] != 'Total', 'Total Engagement'] = engagement_values
results_brand_df.loc[results_brand_df['Brand'] != 'Total', 'Total Keluhan'] = total_keluhan_values

results_brand_df['Rata-rata Engagement'] = results_brand_df['Total Engagement'] / results_brand_df['Total Keluhan'].replace(0, np.nan)
results_brand_df['Rata-rata Engagement'].fillna(0, inplace=True)

total_all_engagement = df['engagement'].sum()
total_terklasifikasi = results_brand_df['Total Engagement'].sum()
total_tidak_terklasifikasi = total_all_engagement - total_terklasifikasi

if total_tidak_terklasifikasi > 0:
    new_row = pd.DataFrame([{
        'Brand': 'Lainnya',
        'Total Engagement': total_tidak_terklasifikasi,
        'Total Keluhan': 0,
        'Rata-rata Engagement': 0,
        'Persentase (%)': (total_tidak_terklasifikasi / total_all_engagement) * 100
    }])
    results_brand_df = pd.concat([results_brand_df, new_row], ignore_index=True)

results_brand_df['Persentase (%)'] = (results_brand_df['Total Engagement'] / total_all_engagement) * 100
results_brand_df['Persentase (%)'].fillna(0, inplace=True)

results_summary_df = results_brand_df[results_brand_df['Brand'] != 'Total'][['Brand', 'Total Engagement', 'Total Keluhan', 'Rata-rata Engagement', 'Persentase (%)']]
display(results_summary_df)
```
