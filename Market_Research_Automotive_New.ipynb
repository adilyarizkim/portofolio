```python
# --- FUNCTION BUILD COMPILED PATTERNS ---
def build_compiled_patterns(use_categories):
    compiled = []
    for category, keywords in use_categories.items():
        for kw in sorted(keywords, key=len, reverse=True):
            pattern = re.compile(r'\b{}\b'.format(re.escape(kw)))
            compiled.append((category, pattern))
    return compiled

# --- COMPILE PATTERN PENGGUNAAN MOBIL ---
compiled_usage_patterns = build_compiled_patterns(use_categories_usage)

# --- COMPILE PATTERN KARAKTERISTIK KONSUMEN ---
compiled_characteristic_patterns = build_compiled_patterns(use_categories_characteristic)

# --- FAST CATEGORY MATCH FUNCTION (GENERAL) ---
def fast_category_match(message, compiled_patterns):
    message_clean = re.sub(r'[^\w\s]', ' ', str(message).lower()).strip()

    for category, pattern in compiled_patterns:
        if pattern.search(message_clean):
            return category

    return None
```

```python
# --- APPLY PENGGUNAAN MOBIL ---
df_matched['Kategori Penggunaan'] = df_matched['message'].apply(lambda x: fast_category_match(x, compiled_usage_patterns))

# --- APPLY KARAKTERISTIK KONSUMEN ---
df_matched['Karakteristik Konsumen'] = df_matched['message'].apply(lambda x: fast_category_match(x, compiled_characteristic_patterns))
```

```python
# --- FILTER YG ADA SALAH SATU KATEGORI (optional) ---
df_matched_categorized = df_matched[
    (df_matched['Kategori Penggunaan'].notnull()) |
    (df_matched['Karakteristik Konsumen'].notnull())
]

def take_3_urls(urls):
    return '; '.join(urls.dropna().head(3))
pivot_final = df_matched_categorized.groupby(
    ['Matched Group', 'Matched Model', 'Matched Brand', 'Kategori Penggunaan', 'Karakteristik Konsumen']
).agg(
    Jumlah_Message = ('message', 'count'),
    Contoh_URL = ('url', take_3_urls)
).reset_index()

print("\n=== PIVOT FINAL: PENGGUNAAN + KARAKTERISTIK + 3 URL ===")
display(pivot_final)
```
