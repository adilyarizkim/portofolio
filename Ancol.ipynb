```python
from google.colab import drive
import pandas as pd
import re

drive.mount('/content/drive')
file_path = '/content/drive/My Drive/Ancol.xlsx'


xls = pd.ExcelFile(file_path)
```

```python
print(xls.sheet_names)
df = xls.parse('Sheet1')
print(df.head())
```

```python
print(f"Jumlah total baris: {len(df)}")
```

```python
keywords_Ancol = [
    "Ancol Taman Impian", "taman impian jaya ancol", "pantai ancol",
    "pantaiancol", "ancoltamanimpian", "tamanimpianjayaancol", "ancol"
]

# Membuat regex pattern dari daftar keyword
pattern_Ancol = '|'.join(re.escape(word) for word in keywords_Ancol)

# Filter berdasarkan kolom 'message'
filtered_ancol = df[df['message'].str.contains(pattern_Ancol, case=False, na=False)]

# Tampilkan jumlah total data yang sesuai dengan keyword
print("Jumlah data yang mengandung keyword Ancol:", len(filtered_ancol))

# Hitung distribusi sentimen
sentiment_count = filtered_ancol['sentiment'].value_counts()
print("\nDistribusi Sentimen:")
print(sentiment_count)
```

Service

```python
# Pastikan kolom 'date' dalam format datetime
filtered_ancol['date'] = pd.to_datetime(filtered_ancol['date'], errors='coerce')

# Hitung jumlah berdasarkan kombinasi tanggal dan service
service_per_day = filtered_ancol.groupby(['date', 'service']).size().reset_index(name='jumlah')

# Buat pivot table
pivot_service = service_per_day.pivot(index='date', columns='service', values='jumlah').fillna(0).astype(int)

# Tambahkan total per hari
pivot_service['total'] = pivot_service.sum(axis=1)

# Tampilkan hasil
print("\n🔹 Jumlah Data per Service per Tanggal:")
display(pivot_service)
```

Sentiment

```python
# Hitung distribusi sentimen
sentiment_count = filtered_ancol['sentiment'].value_counts()
print("\nDistribusi Sentimen:")
print(sentiment_count)
```

```python
# 1. Tampilkan distribusi total sentimen secara keseluruhan
sentiment_count = filtered_ancol['sentiment'].value_counts()
print("🔹 Distribusi Total Sentimen (Keyword: Ancol):")
display(sentiment_count)

# 2. Pastikan kolom 'date' dalam format datetime
filtered_ancol['date'] = pd.to_datetime(filtered_ancol['date'], errors='coerce')

# 3. Filter hanya data dengan sentimen positif dan negatif
filtered_sentiment = filtered_ancol[filtered_ancol['sentiment'].isin(['positive', 'negative'])]

# 4. Hitung jumlah data berdasarkan tanggal dan sentimen
sentiment_per_day = (
    filtered_sentiment
    .groupby(['date', 'sentiment'])
    .size()
    .reset_index(name='jumlah')
)

# 5. Ubah ke bentuk pivot (baris = tanggal, kolom = sentimen)
pivot_sentiment = (
    sentiment_per_day
    .pivot(index='date', columns='sentiment', values='jumlah')
    .fillna(0)
    .astype(int)
)

# 6. Tambahkan kolom total per hari (opsional)
pivot_sentiment['total'] = pivot_sentiment.sum(axis=1)

# 7. Tampilkan hasil
print("\n🔹 Distribusi Harian Sentimen Positif dan Negatif (Keyword: Ancol):")
display(pivot_sentiment)
```

```python
# Salin data hasil filter
df_service_sentiment = filtered_ancol.copy()

# Hitung jumlah kombinasi service dan sentiment
grouped = df_service_sentiment.groupby(['service', 'sentiment']).size().reset_index(name='jumlah')

# Ubah menjadi pivot table: baris = service, kolom = sentiment
pivot_service_sentiment = grouped.pivot(index='service', columns='sentiment', values='jumlah').fillna(0).astype(int)

# Tambahkan kolom total (jumlah seluruh sentimen per service)
pivot_service_sentiment['total'] = pivot_service_sentiment.sum(axis=1)

# Tampilkan hasil akhir
print("Jumlah masing-masing service berdasarkan sentimen dan total data:")
display(pivot_service_sentiment.sort_values(by='total', ascending=False))
```

```python
import re
import pandas as pd

# Keyword Ancol
keywords_Ancol = [
    "Ancol Taman Impian", "taman impian jaya ancol", "pantai ancol",
    "pantaiancol", "ancoltamanimpian", "tamanimpianjayaancol", "ancol"
]

# Keyword per kategori emosi
keywords_Sadness = ["sedih", "nyesek", "merana", "pedih", "pilu", "mewek"]
keywords_Surprise = ["kaget", "terkejut", "heran", "shock", "syok", "kagum", "kejut"]
keywords_Disgust = ["jijik", "mual", "geli", "najis", "risih"]
keywords_Fear = ["takut", "ngeri", "khawatir", "cemas", "merinding", "gelisah"]
keywords_Anger = ["marah", "kesal", "kesel", "emosi", "gila", "muak", "meradang", "jengkel", "ngamuk", "kecewa"]
keywords_Anticipation = ["antisipasi", "waspada", "proyeksi", "gambaran", "awas", "jaga-jaga"]
keywords_Joy = ["senang", "seneng", "bahagia", "semangat", "sejahtera", "betah", "riang"]
keywords_Trust = ["percaya", "ragu", "yakin", "mengakui", "transparan"]

# Filter hanya data dari service selain 'News'
df_filtered = df[df['service'].str.lower() != 'news']

# Filter berdasarkan keyword Ancol
pattern_ancol = '|'.join(re.escape(word) for word in keywords_Ancol)
filtered_ancol = df_filtered[df_filtered['message'].str.contains(pattern_ancol, case=False, na=False)].copy()

# Fungsi deteksi emosi berdasarkan keyword
def detect_emotion(df, column, emotion_keywords, emotion_label):
    pattern = '|'.join(re.escape(word) for word in emotion_keywords)
    mask = df[column].str.contains(pattern, case=False, na=False)
    detected = df[mask].copy()
    detected['emotion'] = emotion_label
    return detected

# Deteksi emosi dari data yang sudah difilter (Ancol only)
emotion_dfs = []
emotion_dfs.append(detect_emotion(filtered_ancol, 'message', keywords_Sadness, 'Sadness'))
emotion_dfs.append(detect_emotion(filtered_ancol, 'message', keywords_Surprise, 'Surprise'))
emotion_dfs.append(detect_emotion(filtered_ancol, 'message', keywords_Disgust, 'Disgust'))
emotion_dfs.append(detect_emotion(filtered_ancol, 'message', keywords_Fear, 'Fear'))
emotion_dfs.append(detect_emotion(filtered_ancol, 'message', keywords_Anger, 'Anger'))
emotion_dfs.append(detect_emotion(filtered_ancol, 'message', keywords_Anticipation, 'Anticipation'))
emotion_dfs.append(detect_emotion(filtered_ancol, 'message', keywords_Joy, 'Joy'))
emotion_dfs.append(detect_emotion(filtered_ancol, 'message', keywords_Trust, 'Trust'))

# Gabungkan semua hasil deteksi emosi
emotion_combined = pd.concat(emotion_dfs, ignore_index=True)

# Hitung jumlah per emosi
total_per_emotion = (
    emotion_combined['emotion']
    .value_counts()
    .reset_index()
    .rename(columns={'index': 'emotion', 'emotion': 'jumlah'})
    .sort_values(by='jumlah', ascending=False)
)

# Tampilkan hasil
print("\n🔹 Total Kemunculan Masing-Masing Emosi (khusus data Ancol, tanpa service 'News'):")
display(total_per_emotion)
```

```python
# Pastikan kolom 'url' tersedia di emotion_combined
print("🔗 Daftar URL dari Data Ancol Berdasarkan Kategori Emosi:\n")

# Loop untuk setiap kategori emosi yang terdeteksi
for emo in sorted(emotion_combined['emotion'].unique()):
    print(f"\n▶️ {emo.upper()} ({(emotion_combined['emotion'] == emo).sum()} posts):")

    # Ambil URL unik dari data dengan emosi tersebut
    urls = (
        emotion_combined[emotion_combined['emotion'] == emo]['url']
        .dropna()
        .unique()
        .tolist()
    )

    if not urls:
        print("Tidak ada URL yang tersedia untuk emosi ini.")
        continue

    # Tampilkan maksimal 10 URL pertama sebagai sampel
    for i, url in enumerate(urls[:10], start=1):
        print(f"{i}. {url}")

    if len(urls) > 10:
        print(f"... dan {len(urls) - 10} URL lainnya")
```

```python
# Hitung ulang engagement berdasarkan kolom yang tersedia
filtered_ancol['engagement'] = (
    filtered_ancol[['likes', 'comments', 'shares', 'views']]
    .fillna(0)
    .sum(axis=1)
)

# Ambil top 5 positif
top_positive = (
    filtered_ancol[filtered_ancol['sentiment'] == 'positive']
    .sort_values(by='engagement', ascending=False)
    .head(5)
    [['message', 'likes', 'comments', 'shares', 'views', 'engagement', 'url']]
)

# Ambil top 5 negatif
top_negative = (
    filtered_ancol[filtered_ancol['sentiment'] == 'negative']
    .sort_values(by='engagement', ascending=False)
    .head(5)
    [['message', 'likes', 'comments', 'shares', 'views', 'engagement', 'url']]
)

# Tampilkan hasil
print("🔹 5 Postingan POSITIF tentang Ancol dengan Engagement Tertinggi:")
display(top_positive)

print("\n🔹 5 Postingan NEGATIF tentang Ancol dengan Engagement Tertinggi:")
display(top_negative)
```
